```python
def verificar_colunas_do_ano(ano):
    caminho = caminhos[ano]
    mapa_ano = mapa_variaveis_por_ano[ano]
    colunas_necessarias = list(mapa_ano.keys())

    colunas_arquivo = pd.read_csv(
        caminho,
        sep=";",
        encoding="latin-1",
        nrows=0
    ).columns.tolist()

    faltantes = [
        coluna for coluna in colunas_necessarias
        if coluna not in colunas_arquivo
    ]

    print(f"Ano {ano}")

    if len(faltantes) == 0:
        print("Todas as colunas necessárias foram encontradas.")
    else:
        print("Colunas faltantes:")
        print(faltantes)
```

função para carregar só as colunas necessárias de cada ano e já renomear para o padrão

```python
def carregar_e_padronizar_microdados(caminho, ano):
    """
    Carrega os microdados de um ano específico e padroniza os nomes das variáveis.
    """

    mapa_ano = mapa_variaveis_por_ano[ano]

    colunas_originais = list(mapa_ano.keys())

    df = pd.read_csv(
        caminho,
        sep=";",
        encoding="latin-1",
        usecols=colunas_originais
    )

    df = df.rename(columns=mapa_ano)

    df["ANO"] = ano

    return df
```

# 3. Filtro geográfico — Nordeste

```python
estados_nordeste = [
    "AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE"
]
```

```python
def filtrar_nordeste(df):
    """
    Filtra apenas participantes vinculados a escolas localizadas no Nordeste.
    """
    df = df[df["UF_ESCOLA"].isin(estados_nordeste)].copy()
    
    return df
```

4. Limpeza dos dados — presença nas provas

```python
def limpar_dados(df):
    """
    Remove treineiros e mantém apenas participantes presentes
    nas quatro provas objetivas do Enem.
    """
    df = df[df["TREINEIRO"] == 0].copy()

    df = df[df["PRESENCA_CN"] == 1].copy()
    df = df[df["PRESENCA_CH"] == 1].copy()
    df = df[df["PRESENCA_LC"] == 1].copy()
    df = df[df["PRESENCA_MT"] == 1].copy()

    return df
```

# 5. Criação da variável dependente — NOTA_MEDIA

```python
def criar_nota_media(df):
    """
    Cria a variável NOTA_MEDIA a partir das quatro provas objetivas do Enem.
    A nota da redação não será considerada.
    """
    df = df.copy()

    df["NOTA_MEDIA"] = df[
        ["NU_NOTA_CN", "NU_NOTA_CH", "NU_NOTA_LC", "NU_NOTA_MT"]
    ].mean(axis=1)

    return df
```

# 6. Variáveis de tecnologia

```python
def criar_variaveis_tecnologia(df):
    """
    Cria variáveis binárias para acesso à internet, computador e celular.
    
    0 = não possui
    1 = possui
    """
    df = df.copy()

    df["INTERNET"] = df["INTERNET_ORIGINAL"].map({
        "A": 0,
        "B": 1
    })

    df["COMPUTADOR"] = df["COMPUTADOR_ORIGINAL"].isin(
        ["B", "C", "D", "E"]
    ).astype(int)

    df["CELULAR"] = df["CELULAR_ORIGINAL"].isin(
        ["B", "C", "D", "E"]
    ).astype(int)

    return df
```

# 7. Variável de renda

```python
def classificar_renda(x):
    """
    Agrupa as faixas de renda familiar em categorias socioeconômicas.
    """

    if x in ["A", "B"]:
        return "classe_baixa_renda"

    elif x == "C":
        return "classe_media_baixa"

    elif x in ["D", "E"]:
        return "classe_media"

    elif x in ["F", "G"]:
        return "classe_media_alta"

    elif x in ["H", "I", "J", "K", "L", "M", "N", "O", "P", "Q"]:
        return "classe_alta_renda"

    else:
        return None
```

```python
def criar_variavel_renda(df):
    """
    Cria a variável RENDA a partir da variável original do questionário.
    """
    df = df.copy()

    df["RENDA"] = df["RENDA_ORIGINAL"].apply(classificar_renda)

    return df
```

# 8. Escolaridade dos pais


```python
def criar_escolaridade_pais(df):
    """
    Cria uma variável única de escolaridade dos pais.

    A variável ESCOLARIDADE_PAIS considera a maior escolaridade informada
    entre pai e mãe.
    """
    df = df.copy()

    ordem_escolaridade = {
        "A": 0,  # Nunca estudou
        "B": 1,  # Não completou a 4ª série/5º ano
        "C": 2,  # Completou a 4ª série/5º ano, mas não completou o fundamental
        "D": 3,  # Completou o fundamental, mas não completou o médio
        "E": 4,  # Completou o médio, mas não completou a faculdade
        "F": 5,  # Completou a faculdade, mas não completou pós-graduação
        "G": 6,  # Completou pós-graduação
        "H": None  # Não sabe
    }

    df["ESCOLARIDADE_PAI"] = df["ESCOLARIDADE_PAI_ORIGINAL"].map(ordem_escolaridade)
    df["ESCOLARIDADE_MAE"] = df["ESCOLARIDADE_MAE_ORIGINAL"].map(ordem_escolaridade)

    df["ESCOLARIDADE_PAIS"] = df[
        ["ESCOLARIDADE_PAI", "ESCOLARIDADE_MAE"]
    ].max(axis=1)

    return df
```

# 9. Tipo de escola

```python
def criar_tipo_escola(df):
    """
    Cria a variável ESCOLA a partir da variável original de tipo de escola.
    
    Para a maior parte dos anos:
    2 = pública
    3 = privada
    
    Em 2018:
    2 = pública
    4 = privada
    """
    df = df.copy()

    def mapear_escola(linha):
        ano = linha["ANO"]
        valor = linha["ESCOLA_ORIGINAL"]

        if ano == 2018:
            mapa = {
                2: "publica",
                4: "privada"
            }
        else:
            mapa = {
                2: "publica",
                3: "privada"
            }

        return mapa.get(valor)

    df["ESCOLA"] = df.apply(mapear_escola, axis=1)

    return df
```

# 10. Base final do modelo

```python
def criar_base_modelo(df):
    """
    Seleciona as variáveis finais do modelo e remove registros
    com valores ausentes nas variáveis de interesse.
    """
    df = df.copy()

    variaveis_modelo = [
        "ANO",
        "NOTA_MEDIA",
        "INTERNET",
        "COMPUTADOR",
        "CELULAR",
        "RENDA",
        "ESCOLA",
        "ESCOLARIDADE_PAIS"
    ]

    df_modelo = df[variaveis_modelo].dropna().copy()

    return df_modelo
```

# 11. Preparação das bases anuais

```python
def preparar_base_ano(ano):
    """
    Aplica todas as etapas de tratamento para um ano específico.
    """
    df = carregar_e_padronizar_microdados(caminhos[ano], ano)

    df = filtrar_nordeste(df)
    df = limpar_dados(df)
    df = criar_nota_media(df)
    df = criar_variaveis_tecnologia(df)
    df = criar_variavel_renda(df)
    df = criar_escolaridade_pais(df)
    df = criar_tipo_escola(df)
    df = criar_base_modelo(df)

    return df
```

```python
fluxo_amostral = []

def registrar_fluxo(ano, etapa, df):
    fluxo_amostral.append({
        "ANO": ano,
        "ETAPA": etapa,
        "PARTICIPANTES": len(df)
    })

def preparar_base_ano_com_fluxo(ano):
    """
    Aplica as etapas de tratamento para um ano específico
    e registra a quantidade de participantes após cada filtro.
    """

    # 1. Carregamento e padronização das variáveis selecionadas
    df = carregar_e_padronizar_microdados(caminhos[ano], ano)
    registrar_fluxo(ano, "Base inicial com variáveis selecionadas", df)

    # 2. Recorte geográfico para escolas do Nordeste
    df = filtrar_nordeste(df)
    registrar_fluxo(ano, "Após recorte para escolas do Nordeste", df)

    # 3. Remoção de treineiros
    df = df[df["TREINEIRO"] == 0].copy()
    registrar_fluxo(ano, "Após remoção de treineiros", df)

    # 4. Presença nas quatro provas objetivas
    df = df[
        (df["PRESENCA_CN"] == 1) &
        (df["PRESENCA_CH"] == 1) &
        (df["PRESENCA_LC"] == 1) &
        (df["PRESENCA_MT"] == 1)
    ].copy()
    registrar_fluxo(ano, "Após presença nas quatro provas objetivas", df)

    # 5. Criação das variáveis utilizadas no modelo
    df = criar_nota_media(df)
    df = criar_variaveis_tecnologia(df)
    df = criar_variavel_renda(df)
    df = criar_escolaridade_pais(df)
    df = criar_tipo_escola(df)

    # 6. Seleção das variáveis finais e remoção de dados ausentes
    df = criar_base_modelo(df)
    registrar_fluxo(ano, "Após dados completos nas variáveis do modelo", df)

    # 7. Remoção de nota média igual a zero
    df = df[df["NOTA_MEDIA"] > 0].copy()
    registrar_fluxo(ano, "Após remoção de nota média igual a zero", df)

    return df
```

```python
fluxo_amostral = []
bases_tratadas_fluxo = []

for ano in anos:
    print(f"Processando ano {ano}...")

    df_ano_fluxo = preparar_base_ano_com_fluxo(ano)

    print(f"Ano {ano}: {df_ano_fluxo.shape[0]} participantes finais")

    bases_tratadas_fluxo.append(df_ano_fluxo)

tabela_fluxo_geral = (
    pd.DataFrame(fluxo_amostral)
    .groupby("ETAPA", sort=False)["PARTICIPANTES"]
    .sum()
    .reset_index()
)

tabela_fluxo_geral["REMOVIDOS_NA_ETAPA"] = (
    tabela_fluxo_geral["PARTICIPANTES"].shift(1) - tabela_fluxo_geral["PARTICIPANTES"]
)

tabela_fluxo_geral["REMOVIDOS_NA_ETAPA"] = (
    tabela_fluxo_geral["REMOVIDOS_NA_ETAPA"]
    .fillna(0)
    .astype(int)
)

tabela_fluxo_geral
```

# 16. Modelo de regressão linear OLS

```python
df_longitudinal["RENDA"].value_counts()
```

```python
df_longitudinal["ESCOLA"].value_counts()
```

```python
modelo_ols_robusto = smf.ols(
    formula="""
        NOTA_MEDIA ~ INTERNET
                   + COMPUTADOR
                   + CELULAR
                   + C(RENDA, Treatment(reference="classe_baixa_renda"))
                   + C(ESCOLA, Treatment(reference="publica"))
                   + ESCOLARIDADE_PAIS
                   + C(ANO)
    """,
    data=df_longitudinal
).fit(cov_type="HC3")

print(modelo_ols_robusto.summary())
```

```python
resultado_modelo = pd.DataFrame({
    "coeficiente": modelo_ols_robusto.params,
    "erro_padrao": modelo_ols_robusto.bse,
    "p_valor": modelo_ols_robusto.pvalues
})

resultado_modelo
```

# 17 Tabela resumida dos principais resultados do modelo

```python
variaveis_principais = [
    "INTERNET",
    "COMPUTADOR",
    "CELULAR",
    "ESCOLARIDADE_PAIS",
    'C(ESCOLA, Treatment(reference="publica"))[T.privada]',
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_media_baixa]',
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_media]',
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_media_alta]',
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_alta_renda]'
]

tabela_modelo = resultado_modelo.loc[variaveis_principais].copy()

tabela_modelo = tabela_modelo.round({
    "coeficiente": 3,
    "erro_padrao": 3,
    "p_valor": 4
})

tabela_modelo
```

# 18. Tabela resumida dos principais resultados do modelo

Após a estimação do modelo OLS com erros-padrão robustos, foi construída uma tabela resumida contendo os principais coeficientes de interesse da pesquisa. Essa tabela reúne as variáveis relacionadas aos recursos digitais, além das principais variáveis de controle utilizadas no modelo.

```python
tabela_modelo_formatada = tabela_modelo.reset_index()

tabela_modelo_formatada = tabela_modelo_formatada.rename(columns={
    "index": "Variável",
    "coeficiente": "Coeficiente",
    "erro_padrao": "Erro padrão",
    "p_valor": "P-valor"
})

tabela_modelo_formatada["Variável"] = tabela_modelo_formatada["Variável"].replace({
    "INTERNET": "Acesso à internet",
    "COMPUTADOR": "Possui computador",
    "CELULAR": "Possui celular",
    "ESCOLARIDADE_PAIS": "Escolaridade dos pais",
    'C(ESCOLA, Treatment(reference="publica"))[T.privada]': "Escola privada",
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_media_baixa]': "Classe média baixa",
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_media]': "Classe média",
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_media_alta]': "Classe média alta",
    'C(RENDA, Treatment(reference="classe_baixa_renda"))[T.classe_alta_renda]': "Classe alta renda"
})

tabela_modelo_formatada
```
